### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ingreso_total_tienda1 = tienda['Precio'].sum()
ingreso_total_tienda2 = tienda2['Precio'].sum()
ingreso_total_tienda3 = tienda3['Precio'].sum()
ingreso_total_tienda4 = tienda4['Precio'].sum()

print(f"Ingreso total Tienda 1: {ingreso_total_tienda1:,.2f}")
print(f"Ingreso total Tienda 2: {ingreso_total_tienda2:,.2f}")
print(f"Ingreso total Tienda 3: {ingreso_total_tienda3:,.2f}")
print(f"Ingreso total Tienda 4: {ingreso_total_tienda4:,.2f}")
```

# 4. Productos más y menos vendidos

```python
product_counts_tienda1 = tienda['Producto'].value_counts()

most_sold_product_tienda1 = product_counts_tienda1.idxmax()
least_sold_product_tienda1 = product_counts_tienda1.idxmin()

print(f"Producto más vendido (Tienda 1): {most_sold_product_tienda1} ({product_counts_tienda1.max()} ventas)")
print(f"Producto menos vendido (Tienda 1): {least_sold_product_tienda1} ({product_counts_tienda1.min()} ventas)")
```

```python
product_counts_tienda2 = tienda2['Producto'].value_counts()

most_sold_product_tienda2 = product_counts_tienda2.idxmax()
least_sold_product_tienda2 = product_counts_tienda2.idxmin()

print(f"Producto más vendido (Tienda 2): {most_sold_product_tienda2} ({product_counts_tienda2.max()} ventas)")
print(f"Producto menos vendido (Tienda 2): {least_sold_product_tienda2} ({product_counts_tienda2.min()} ventas)")
```

```python
product_counts_tienda3 = tienda3['Producto'].value_counts()

most_sold_product_tienda3 = product_counts_tienda3.idxmax()
least_sold_product_tienda3 = product_counts_tienda3.idxmin()

print(f"Producto más vendido (Tienda 3): {most_sold_product_tienda3} ({product_counts_tienda3.max()} ventas)")
print(f"Producto menos vendido (Tienda 3): {least_sold_product_tienda3} ({product_counts_tienda3.min()} ventas)")
```

```python
product_counts_tienda4 = tienda4['Producto'].value_counts()

most_sold_product_tienda4 = product_counts_tienda4.idxmax()
least_sold_product_tienda4 = product_counts_tienda4.idxmin()

print(f"Producto más vendido (Tienda 4): {most_sold_product_tienda4} ({product_counts_tienda4.max()} ventas)")
print(f"Producto menos vendido (Tienda 4): {least_sold_product_tienda4} ({product_counts_tienda4.min()} ventas)")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


most_and_least_sold_data = {
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Producto Más Vendido': [
        most_sold_product_tienda1,
        most_sold_product_tienda2,
        most_sold_product_tienda3,
        most_sold_product_tienda4
    ],
    'Ventas Más Vendido': [
        product_counts_tienda1.max(),
        product_counts_tienda2.max(),
        product_counts_tienda3.max(),
        product_counts_tienda4.max()
    ],
    'Producto Menos Vendido': [
        least_sold_product_tienda1,
        least_sold_product_tienda2,
        least_sold_product_tienda3,
        least_sold_product_tienda4
    ],
    'Ventas Menos Vendido': [
        product_counts_tienda1.min(),
        product_counts_tienda2.min(),
        product_counts_tienda3.min(),
        product_counts_tienda4.min()
    ]
}

summary_df = pd.DataFrame(most_and_least_sold_data)
print(summary_df)
```

```python
plt.figure(figsize=(15, 6))

# Plot for most sold products
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
sns.barplot(x='Tienda', y='Ventas Más Vendido', hue='Producto Más Vendido', data=summary_df, palette='viridis')
plt.title('Productos Más Vendidos por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Cantidad de Ventas')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Producto', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Plot for least sold products
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
sns.barplot(x='Tienda', y='Ventas Menos Vendido', hue='Producto Menos Vendido', data=summary_df, palette='magma')
plt.title('Productos Menos Vendidos por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Cantidad de Ventas')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Producto', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

plt.show()
```

## 1. Introducción

El presente informe tiene como objetivo analizar el desempeño de las cuatro tiendas de **Alura Store Latam** con el fin de recomendar en cuál de ellas debería vender el Sr. Juan.

Para ello, se evaluaron los siguientes factores clave:

* Ingresos totales por tienda
* Categorías de productos más y menos vendidas
* Calificación promedio de los clientes
* Productos más y menos vendidos
* Costo de envío promedio

El análisis combina indicadores financieros, operativos y de satisfacción del cliente, permitiendo una decisión estratégica basada en datos.

---

## 2. Desarrollo del Análisis

### 2.1 Ingresos Totales

*   Ingreso total Tienda 1: 1,150,880,400.00

*   Ingreso total Tienda 3: 1,098,019,600.00
*   Ingreso total Tienda 4: 1,038,375,700.00

* La tienda con mayor ingreso es la Tienda 1

**Análisis:**
La **Tienda 1** presenta el mayor nivel de facturación, lo que indica alto volumen de ventas y fuerte posicionamiento comercial. Sin embargo, la diferencia con Tienda 2 y Tienda 3 no es extremadamente amplia, por lo que es necesario evaluar otros factores.

---

### 2.2 Categorías de Productos Más Vendidas

**Muebles**

| Factor                 | Mejor Tienda |

## 4. Conclusión y Recomendación

Después de analizar todos los factores estratégicos, se recomienda que **el Sr. Juan venda en la Tienda 3**.

Si el Sr. Juan busca una decisión sostenible a largo plazo, basada no solo en ingresos actuales sino en reputación y potencial de crecimiento, **Tienda 3 es la mejor opción**.
